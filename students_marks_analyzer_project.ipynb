```python
#Project: Student Marks Analyzer using numpy and python...

import numpy as np

marks = np.array([85, 72, 90, 45, 60, 35, 95])

# Total marks
total = np.sum(marks)

# Average marks
average = np.mean(marks)

# Highest marks
highest = np.max(marks)

# Lowest marks
lowest = np.min(marks)

# Pass students
passed = marks[marks >= 40]

print("Total Marks :", total)
print("Average Marks :", average)
print("Highest Marks :", highest)
print("Lowest Marks :", lowest)
print("Passed Marks :", passed)
```
